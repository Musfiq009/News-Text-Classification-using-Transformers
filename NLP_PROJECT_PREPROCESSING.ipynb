# Import Libraries

```python
import pandas as pd
import numpy as np
```

# Load Dataset to DataFrame

```python
df1 = pd.read_csv('/content/drive/MyDrive/bbc-text.csv')
df2 = pd.read_csv('/content/drive/MyDrive/News-Categoires.csv', encoding='latin1')
```

# Display Dataset-1

```python
# Dataset 1: BBC News - take first 100 rows
#df1 = df1.head(100)
print("DATASET 1 - BBC News")
print("Shape:", df1.shape)
print("Columns:", df1.columns.tolist())
print(df1.head())
print(df1['category'].value_counts())
print()
```

# Case Folding

```python
text_samples = df1['text'].tolist()
normalized_texts = [x.lower() for x in text_samples]
df1['lower'] = normalized_texts
print(df1['lower'].head())
```

# Punctuation Removal

```python
import string

print(string.punctuation)

def remove_punctuation(text):
    text_remove_mapping = text.maketrans("", "", string.punctuation)
    text = text.translate(text_remove_mapping)
    return text.strip()

df1['punctuation_remove'] = df1['lower'].apply(remove_punctuation)
print(df1['punctuation_remove'].head())
```

# Tokenization

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')

df1['tokenization'] = df1['punctuation_remove'].apply(word_tokenize)
print(df1['tokenization'].head())
```

# Stopward Removal

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words("english"))

def remove_stopwords(tokens):
    filtered_words = [word for word in tokens if word not in stop_words]
    return filtered_words

df1['stopword_removal'] = df1['tokenization'].apply(remove_stopwords)
print(df1['stopword_removal'].head())
```

# Lematization

```python
import spacy
nlp = spacy.load("en_core_web_sm")

def lemmatize_tokens(tokens):
    doc = nlp(" ".join(tokens))
    return [token.lemma_ for token in doc]

df1['lemmatized'] = df1['stopword_removal'].apply(lemmatize_tokens)
print(df1['lemmatized'].head())
```

```python
df1['clean_text'] = df1['lemmatized'].apply(lambda tokens: " ".join(tokens))
print(df1[['text', 'clean_text']].head())
```

```python
df1[['category', 'clean_text']].to_csv('/content/drive/MyDrive/bbc-text-preprocessed.csv', index=False)
```
